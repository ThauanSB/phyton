```python
#Exercicio 9
#Um professor gostaria um programa para auxiliá-lo a montar a média final de seus alunos.
#Sabendo que são 2 notas por ano, monte um programa que através das notas informadas
#pelo usuário mostre a sua média final. Não esqueça de manter uma boa interface com o
#usuário!!
print('Calculo de media escolar digite a primeira e a segunda nota para ter a media:')
num1 = float(input('Primeira Nota:'))
num2 = float(input('Segunda Nota:'))
mediaSoma = (num1 + num2) / 2
print('A media do aluno e:' + str(mediaSoma))
```

```python
#Exercicio 10
#Faça um programa para converter Reais em Dólares informado pelo usuário:
print('Informe o valor e a cotaçao atual do Dolar:')
num1 = float(input('Valor em Reais:'))
num2 = float(input('Cotaçao atual do Dolar:'))
numSoma = round(num1 / num2, 2)
print('O seu valor em Dolar e:' + str(numSoma))
```

```python
#Exercicio 11
#Faça um programa para converter Graus Celsius em Fahrenheit informado pelo usuário:
print('Informe os graus Celsius para converçao em Fahrenheit:')
num1 = float(input('A temperatura atual:'))
valorFahrenheit = (num1 * 1.8) + int(32)
print('A temperatura atual em Fahrenheit e :' + str(valorFahrenheit))
```
